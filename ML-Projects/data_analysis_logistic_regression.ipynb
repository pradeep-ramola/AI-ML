Handle missing entries. Possible solutions:
• Drop rows with missing entries. If you have a large dataset and only a few missing
features, it may be acceptable to drop the rows containing missing values.
• Impute missing data. Replace the missing entries with the mean/median/mode of the
feature. You can use the k-nearest neighbor algorithm to find the matching sample.

```python
df_clean = df
# For the numerical columns we will get the mean of each column and fill the null values with that mean
# print(df_clean.select_dtypes(include=np.number).columns)
for col in df_clean.select_dtypes(include=np.number).columns:
    mean_val = df_clean[col].mean()
    df_clean[col].fillna(mean_val, inplace=True)


# for categorical data we can either use the most repeated values [mode] or we can use knn or decision tree
# Let's go with the mode

for col in df_clean.select_dtypes(include='object').columns:
    mode_val = df_clean[col].mode()[0]
    df_clean[col].fillna(mode_val, inplace=True)

```

Step 2: Refactor into a Reusable Preprocessing Method [3 points]

• This can be a class (e.g., PreprocessPipeline) or a clean set of functions.
• Your code should include:
o Handling missing values
o Cleaning string formats
o Outlier detection and treatment
o Encoding categorical variables
o Scaling numeric features

```python


class PreprocessingPipeline:

    def __init__(self, numerical_cols, categorical_cols, missing_num_strategy='mean', outlier_method='remove', iqr_multiplier=1.5):
        self.numerical_cols = numerical_cols
        self.categorical_cols = categorical_cols
        self.missing_num_strategy = missing_num_strategy
        self.outlier_method = outlier_method
        self.iqr_multiplier = iqr_multiplier
        self.imputation_values = {}
        self.scaling_params = {}
        self.outlier_bounds = {}
        self.columns_to_encode = []


    def fit(self, df):
        df_fit = df.copy()

        for col in self.numerical_cols:
            if self.missing_num_strategy == 'mean':
                self.imputation_values[col] = df_fit[col].mean()
            # elif self.missing_num_strategy == 'median':
            #     self.imputation_values[col] = df_fit[col].median()
        for col in self.categorical_cols:
            self.imputation_values[col] = df_fit[col].mode()[0]

        #outlier from numerical columns
        for col in self.numerical_cols:
            Q1 = df_fit[col].quantile(0.25)
            Q3 = df_fit[col].quantile(0.75)
            IQR = Q3 - Q1
            lower_bound = Q1 - self.iqr_multiplier * IQR
            upper_bound = Q3 + self.iqr_multiplier * IQR
            self.outlier_bounds[col] = {'lower': lower_bound, 'upper': upper_bound}

        #columns for encoding
        self.columns_to_encode = self.categorical_cols

        #scaling parameters min/max from numerical columns
        for col in self.numerical_cols:
            self.scaling_params[col] = {'min': df_fit[col].min(), 'max': df_fit[col].max()}

        return self

    def transform(self, df):
        df_transformed = df.copy()

        #Handle missing values using learned parameters
        df_transformed.fillna(self.imputation_values, inplace=True)

        #string formats
        for col in self.categorical_cols:
            df_transformed[col] = df_transformed[col].astype(str).str.capitalize()


        #Handle outliers
        if self.outlier_method == 'remove':
            keep_mask = pd.Series(True, index=df_transformed.index)
            for col in self.numerical_cols:
                lower = self.outlier_bounds[col]['lower']
                upper = self.outlier_bounds[col]['upper']
                keep_mask &= (df_transformed[col] >= lower) & (df_transformed[col] <= upper)
            df_transformed = df_transformed[keep_mask]

        #encode categorical variables
        df_transformed = pd.get_dummies(df_transformed, columns=self.columns_to_encode, drop_first=True)

        #Scale numeric features
        for col in self.numerical_cols:
            min_val = self.scaling_params[col]['min']
            max_val = self.scaling_params[col]['max']
            if (max_val - min_val) != 0:
                df_transformed[col] = (df_transformed[col] - min_val) / (max_val - min_val)
            else:
                df_transformed[col] = 0
        print("Data transformation complete.")
        return df_transformed

    def fit_transform(self, df):
        self.fit(df)
        return self.transform(df)

    def get_summary(self):

        print("=====Preprocessing Pipeline Settings Summary====")
        print(f"Numerical Features: {self.numerical_cols}")
        print(f"Categorical Features: {self.categorical_cols}")

        print("Step 1: Handling Missing Value ")
        print(f"Numerical Strategy: '{self.missing_num_strategy}'")
        print(f"Learned Imputation Values:")
        for col, val in self.imputation_values.items():
            if isinstance(val, (int, float)):
                print(f"    - {col}: {val:.2f}")
            else:
                print(f"    - {col}: {val}")

        print("Step 2: String Format ")
        print("  - Strategy: Convert all categorical features to capitalized format.")

        print("Step 3: Outlier Handling")
        print(f"  - Method: '{self.outlier_method}'")
        print(f"  - IQR Multiplier: {self.iqr_multiplier}")

        print("Step 4: Categorical Encoding")
        print(f"  - Method: One-Hot Encoding (with 'drop_first=True')")
        print(f"  - Features Encoded: {self.columns_to_encode}")

        print("Step 5: Numeric Feature Scaling")
        print(f"  - Method: Min-Max Normalization (to 0-1 range)")
        print(f"  - Learned Min/Max Values (from original data):")
        for col, params in self.scaling_params.items():
            print(f"    - {col}: Min={params['min']:.2f}, Max={params['max']:.2f}")
        print("\n" + "="*50)
```

Now let's split the data for training


```python


# Apply Min-Max scaling
X= second_penguin.drop(target_column, axis=1)
X_normalized = (X - X.min()) / (X.max() - X.min())


X_normalized = X_normalized.fillna(0)


X = X_normalized.values

print(X_normalized.head())
```

```python

def split_data(X, Y, test_size=0.2, random_state=42):

    np.random.seed(random_state)

    #number of samples
    n_samples = X.shape[0]


    random_indices = np.random.permutation(n_samples)


    set_size = int(n_samples * test_size)


    test_indices = random_indices[:set_size]
    train_indices = random_indices[set_size:]

    # Split the data using the indices
    X_train, X_test = X[train_indices, :], X[test_indices, :]
    y_train, y_test = Y[train_indices, :], Y[test_indices, :]

    return X_train, X_test, y_train, y_test


X_train, X_test, y_train, y_test = split_data(X, Y, test_size=0.2, random_state=42)

print("Train/Test Shapes")
print(f"X_train shape: {X_train.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_test shape: {y_test.shape}")


```

```python


class LogisticRegression():
    def __init__(self, learning_rate, iterations):

        self.learning_rate = learning_rate
        self.iterations = iterations
        self.weights = None
        self.bias = None
        self.loss_history = []

    def _sigmoid(self, z):
        # Formula: σ(z) = 1 / (1 + e^(-z))
        # refrence - https://www.pythontutorials.net/blog/numpy-clamp/
        z = np.clip(z, -1000, 1000)
        return 1 / (1 + np.exp(-z))

    def _cost(self, y, y_hat):

        #https://pythonandml.github.io/dlbook/content/multilayer_perceptrons/cost_functions.html
        #adding epilson as small value to avoid log0
        e = 1e-9
        cost = -np.mean(y * np.log(y_hat + e) + (1 - y) * np.log(1 - y_hat + e))
        return cost


    def fit(self, X, y):
            #  using gradient descent.

        #samples and features
        n_samples, n_features = X.shape

        # Initialize weights randomly with small values and bias to 0.01
        self.weights = np.random.uniform(0, 0.01, (n_features, 1))
        self.bias = 0.01
        self.loss_history = []


        for i in range(self.iterations):
            #Gradient Descent Steps
            #1.Calculate using (z = w^T*X + b)
            z = np.dot(X, self.weights) + self.bias

            #2 get prediction (y_hat) using the sigmoid function
            y_hat = self._sigmoid(z)

            #calculate the cost using loss function
            cost = self._cost(y, y_hat)
            self.loss_history.append(cost)

            #loss function derivate - gradient function
            # dJ/dw = 1/N * X^T * (y_hat - y)
            dw = (1 / n_samples) * np.dot(X.T, (y_hat - y))
            # dJ/db = 1/N * Σ(y_hat - y)
            db = (1 / n_samples) * np.sum(y_hat - y)

            #Updating weights and bias
            self.weights -= self.learning_rate * dw
            self.bias -= self.learning_rate * db


            if (i % 10000 == 0) or (i == self.iterations - 1):
                print(f"Iteration {i}: Loss = {cost:.6f}")

        print("Training complete.")

    def predict(self, X):
        """
        Makes binary predictions (0 or 1) for a given input X.
        """
        # Calculate according to the formula y=mx+c
        z = np.dot(X, self.weights) + self.bias
        # Get the probabilities using sigmoid / sign activation
        probabilities = self._sigmoid(z)

        # Convert probabilities to binary class labels
        # Prediction is 1 if probability >= 0.5, otherwise 0
        y_pred_binary = (probabilities >= 0.5).astype(int)

        return y_pred_binary




#Hyperparameters
hyperparameter_setups = [
    {"name": "(Balanced LR)", "learning_rate": 0.001, "iterations": 100000},
    {"name": "(High LR)", "learning_rate": 0.02, "iterations": 20000},
    {"name": "(Low LR)", "learning_rate": 0.0001, "iterations": 150000}
]

#store for plotting and analysis
trained_models = []


for setup in hyperparameter_setups:
    print("\n" + "="*50)
    print(f"--- Training Model with {setup['name']} ---")
    print(f"Learning Rate: {setup['learning_rate']}, Iterations: {setup['iterations']}")


    model = LogisticRegression(learning_rate=setup['learning_rate'], iterations=setup['iterations'])

    # Train the model using the training data
    model.fit(X_train, y_train)

    # Store the trained model and its settings for evaluation
    trained_models.append({"model": model, "name": setup['name']})
```

```python
def calculate_metrics(y_true, y_pred):

    # Calculate the four components of the confusion matrix
    TP = np.sum((y_true == 1) & (y_pred == 1)) # True Positives
    TN = np.sum((y_true == 0) & (y_pred == 0)) # True Negatives
    FP = np.sum((y_true == 0) & (y_pred == 1)) # False Positives
    FN = np.sum((y_true == 1) & (y_pred == 0)) # False Negatives

    #Calculate metrics and handle division by 0 if its 0 just keep the ans as 0
    total = TP + TN + FP + FN
    accuracy = (TP + TN) / total if total > 0 else 0
    precision = TP / (TP + FP) if (TP + FP) > 0 else 0
    recall = TP / (TP + FN) if (TP + FN) > 0 else 0
    f1_score = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

    metrics = {
        "TP": TP, "TN": TN, "FP": FP, "FN": FN,
        "accuracy": accuracy, "precision": precision,
        "recall": recall, "f1_score": f1_score
    }
    return metrics

# Looping through our trained model
for item in trained_models:
    model = item['model']
    name = item['name']

    #predict the y value
    y_pred_test = model.predict(X_test)

    #call the method by passing target and predicted
    metrics = calculate_metrics(y_test, y_pred_test)

    #results
    print("\n" + "="*100)
    print(f"Performance Evaluation={name}")
    print(f"Accuracy:  {metrics['accuracy']:.6f}")
    print(f"Precision: {metrics['precision']:.5f}")
    print(f"Recall:    {metrics['recall']:.6f}")
    print(f"F1-Score:  {metrics['f1_score']:.5f}")

    #Confusion matrix
    print("==================================")
    print("                Machine Negative | Machine Positive")
    print("               --------------------------------------")
    print(f"Actual Negative |      {metrics['TN']:<10} |      {metrics['FP']:<10}")
    print(f"Actual Positive |      {metrics['FN']:<10} |      {metrics['TP']:<10}")
```

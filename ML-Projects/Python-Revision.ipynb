4. Counting occurrences

Write a function that counts the number of times a given word appears in a sentence.

The word should be matched as a whole word, and the search should be case-insensitive.

For example: "The book is on the table", total occurances of "the" is 2.

Reflection: Why is case-insensitive matching often needed in text processing?

````python
import string


sentence = "The cat and the dog played in the park next to the theater. The cat was happy."
word_to_count = "The"


# def CountOccurrence(sentence:str, wordToCount:str)->int:
#     if not sentence or not wordToCount:
#         return 0
#     return sentence.lower().strip(string.punctuation).split().count(wordToCount)


# print(CountOccurrence(sentence,word_to_count))
# doesn't handle edge cases of punctuation and . what if our word to count was just theatre it comes with period with this operation
# BEGIN_YOUR_CODE
def CountOccurrence(sentence:str, wordToCount:str)->int:
    if not sentence or not word_to_count:
        return 0

    wordDict={}

    words=sentence.lower().split(" ")

    for word in words:
        word=word.strip(string.punctuation)
        wordDict[word]= wordDict.get(word,0)+1

    return wordDict.get(wordToCount.lower(),0)
print(f"The occurrence of word {word_to_count} in the sentence is {CountOccurrence(sentence,word_to_count)} \n")

print("Reflection why case insensitive matters in text processing")
print("Because words can be in different case such as all capital or all lower or just a mixture of both")
print("Such as The the and THE all are same so we inorder to compare or search or like our example of word count. we need to bring the words in uniform state")
print("we can either capitalize all the letters in a word or to all lower inorder to compare the words effectively")
print("Computer only understand binary in the end The the and THE all are same")
print("Case insensitive comparison helps us to overcome the challenges comparing , searching or keeping the word count")
# END_YOUR_CODE
````

### Euclidian Distance

Let point $p$ have Cartesian coordinates $(p_1,p_2)$ and let point $q$ have coordinates $(q_1,q_2)$. Then the distance between $p$ and $q$ is given by:
    $d(p,q) = \sqrt{(q_1-p_1)^2 + (q_2-p_2)^2}$.</math>

1. Euclidean distance
Using numpy library return the Euclidean distance between two locations, where the locations are pairs of numbers (e.g., (3, 5))

````python
def euclidean_distance(loc1: int, loc2: int) -> float:


    # BEGIN_YOUR_CODE
    value1= np.array(loc1)
    value2= np.array(loc2)
    difference = value2-value1
    sumSquare= np.sum(difference**2)
    return np.sqrt(sumSquare)
    # raise Exception("Not implemented yet")

    # END_YOUR_CODE
````

3. Fit the model
- Create and fit a logistic regression model to predict the 'Species' of a penguin using all of the other features in the dataset.
- You can use one hot encoding to encode categorical-based feature ('island' and 'gender') or reuse your code from abouve
- Create a logistic regression model using LogisticRegression().
- Fit the model using the training data
- Make predictions on the testing data using the trained model
- Print the accuracy score by comparing the predicted values with the actual target values using accuracy_score()
- Print the parameter values of the logistic regression model using get_params()

You can read more about the inbuild Logistic Regression model here - https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LogisticRegression.html

````python
# BEGIN_YOUR_CODE

from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score


X = cleaned_df[["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g", "island", "gender"]]
Y = cleaned_df["species"]

# splitting data
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=40, stratify=Y
)

categorical = ["island", "gender"]
numeric = ["flipper_length_mm","bill_depth_mm", "bill_length_mm","body_mass_g"]

# Reference- https://medium.com/@bharkristanto/classfication-with-penguins-dataset-using-logistic-regression-9b51703d5e65
preprocessor = ColumnTransformer(
    transformers=[
        ("categorical", OneHotEncoder(drop="first"), ["island", "gender"]),
        ("numeric", "passthrough", ["bill_length_mm","bill_depth_mm","flipper_length_mm","body_mass_g"])
    ]
)



model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000))
])


model.fit(X_train, Y_train)


Y_pred = model.predict(X_test)


acc = accuracy_score(Y_test, Y_pred)
print("Accuracy is: \n", acc)


print("Logistic Regression Parameters: \n")
print(model.named_steps["classifier"].get_params())


# END_YOUR_CODE
````

2. Optimize hyperparameters using Optuna
- Reuse the code from Part 5 for fitting a logistic regression model using scikit-learn
- Use Optuna to determine the optimal values for two hyperparameters in logistic regression: C and max_iter.

    The aim is to find the hyperparameters that result in the highest accuracy score
    
    The hyperparameter C should be in the range [0.01, 10], and max_iter should be within the range of [100, 1000]
- Print the optimal values of C and max_iter and corresponding accuracy score obtained

````python
# BEGIN_YOUR_CODE
def objective(trial):
    C = trial.suggest_float("C", 0.01, 10.0)
    maxIteration = trial.suggest_int("maxIteration", 100, 1000)

    model = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(C=C, max_iter=maxIteration, random_state=50))
    ])

    model.fit(X_train, Y_train)
    Y_pred = model.predict(X_test)
    acc = accuracy_score(Y_test, Y_pred)

    return acc


study = op.create_study(direction="maximize")
study.optimize(objective, n_trials=40, show_progress_bar=True)


print("Best Hyperparameters are:")
print("Maximum iteration:", study.best_params["maxIteration"])
print("C:", study.best_params["C"])
print("Best Accuracy:", study.best_value)


# END_YOUR_CODE
````
